```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris


iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df['species'] = iris.target

print("first 5 rows : ")
print(df.head())
print("\n")

print("dataset shape : ")
print(df.shape)
print("\n")

print("column name:")
print(df.columns)
print("\n")

print("data info:")
df.info()

print("\n")

print("Stats Summary ")
print(df.describe())

df.hist(figsize=(10, 6), bins=10, edgecolor='black')
plt.tight_layout()
plt.show()
```
