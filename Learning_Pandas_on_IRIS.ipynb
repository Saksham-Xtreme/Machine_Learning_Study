```python
# Import Pandas for working with data in the form of DataFrames
import pandas as pd

# Import Matplotlib for creating graphs and visualizations
import matplotlib.pyplot as plt

# Import the built-in Iris dataset from Scikit-learn
from sklearn.datasets import load_iris


# ---------------------------------------------------------
# LOAD THE DATASET
# ---------------------------------------------------------

# Load the Iris dataset
# The returned object contains:
# - data          -> input/features
# - target        -> numerical labels
# - feature_names -> names of the features
# - target_names  -> actual species names
iris = load_iris()


# ---------------------------------------------------------
# CREATE DATAFRAME
# ---------------------------------------------------------

# Convert the feature data into a Pandas DataFrame
# iris.data contains 4 features for every flower:
# 1. sepal length
# 2. sepal width
# 3. petal length
# 4. petal width
#
# iris.feature_names provides the names for these columns
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)


# Add the target/species values as a new column
#
# The target values are numerical:
# 0 -> Setosa
# 1 -> Versicolor
# 2 -> Virginica
df['species'] = iris.target


# ---------------------------------------------------------
# VIEW FIRST 5 ROWS
# ---------------------------------------------------------

print("First 5 rows:")

# head() displays the first 5 rows of the DataFrame
# It is commonly used to quickly inspect a dataset
print(df.head())

print("\n")


# ---------------------------------------------------------
# DATASET SHAPE
# ---------------------------------------------------------

print("Dataset shape:")

# shape returns:
# (number_of_rows, number_of_columns)
#
# Iris dataset:
# 150 rows and 5 columns
print(df.shape)

print("\n")


# ---------------------------------------------------------
# COLUMN NAMES
# ---------------------------------------------------------

print("Column names:")

# columns returns the names of all columns in the DataFrame
print(df.columns)

print("\n")


# ---------------------------------------------------------
# DATASET INFORMATION
# ---------------------------------------------------------

print("Data info:")

# info() gives important information about the DataFrame:
# - number of rows
# - column names
# - number of non-null values
# - data types
# - memory usage
#
# It is useful for understanding the structure of the dataset
df.info()

print("\n")


# ---------------------------------------------------------
# STATISTICAL SUMMARY
# ---------------------------------------------------------

print("Stats Summary:")

# describe() generates basic statistical information
# for numerical columns:
#
# count -> number of values
# mean  -> average
# std   -> standard deviation
# min   -> minimum value
# 25%   -> first quartile
# 50%   -> median
# 75%   -> third quartile
# max   -> maximum value
print(df.describe())

print("\n")


# ---------------------------------------------------------
# HISTOGRAM
# ---------------------------------------------------------

# hist() creates histograms for numerical columns
#
# A histogram shows how values are distributed.
# For example, it helps us understand how many flowers
# have a particular sepal length.
#
# figsize=(10, 6) -> width = 10, height = 6
# bins=10 -> divide the values into 10 intervals
# edgecolor='black' -> add borders around histogram bars
df.hist(
    figsize=(10, 6),
    bins=10,
    edgecolor='black'
)


# Automatically adjust spacing between the graphs
# so that labels and titles do not overlap
plt.tight_layout()


# Display the histogram
plt.show()


print("\n")


# ---------------------------------------------------------
# SPECIES COUNT BAR CHART
# ---------------------------------------------------------

# value_counts() counts how many times each species value occurs
#
# Example:
# 0 -> 50 flowers
# 1 -> 50 flowers
# 2 -> 50 flowers
#
# sort_index() sorts the species values:
# 0, 1, 2
#
# plot(kind='bar') converts these counts into a bar chart
df['species'].value_counts().sort_index().plot(
    kind='bar',

    # Give each species a different color
    color=['skyblue', 'orange', 'green'],

    # Add black borders around the bars
    edgecolor='black'
)


# Add a title to the bar chart
plt.title('Species Count')

# Label the X-axis
plt.xlabel('Species')

# Label the Y-axis
plt.ylabel('Count')


# Display the bar chart
plt.show()

print("\n")


# ---------------------------------------------------------
# SCATTER PLOT
# ---------------------------------------------------------

# Create a new figure for the scatter plot
#
# figsize=(7, 5) means:
# width = 7 inches
# height = 5 inches
plt.figure(figsize=(7, 5))


# Create a scatter plot
#
# X-axis:
# Sepal length
#
# Y-axis:
# Sepal width
#
# c=df['species']:
# Color each point according to its species
#
# cmap='viridis':
# Use the Viridis color map for the species colors
plt.scatter(
    df['sepal length (cm)'],
    df['sepal width (cm)'],
    c=df['species'],
    cmap='viridis'
)


# Add title to the scatter plot
plt.title('Sepal Length vs Sepal Width')

# Label the X-axis
plt.xlabel('Sepal Length (cm)')

# Label the Y-axis
plt.ylabel('Sepal Width (cm)')


# Display the scatter plot
plt.show()
```
