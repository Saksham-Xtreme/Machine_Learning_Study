```
import matplotlib.pyplot as plt

names= ["Aman", "Ravi", "Priya", "Neha", "Rahul", "Simran",
        "Karan", "Anjali", "Vikas", "Pooja"]

marks= [70, 85, 92, 67, 74, 89, 95, 81, 69, 90]

# ------ Line Plot -------

plt.figure(figsize= (10, 5))

plt.plot(names, marks, marker= 'o', color= "red", linewidth= 2)
plt.xlabel("Names")
plt.ylabel("Marks")
plt.title("Marks of 10 Persons - Line Plot")
plt.grid(True)
plt.show()


# ------- Bar Plot ---------
print("\n")
plt.figure(figsize= (10, 5))

plt.bar(names, marks, color= "pink")
plt.xlabel("Names")
plt.ylabel("Marks")
plt.title("Marks of 10 Persons - Bar Plot")
plt.grid(True)
plt.show()

#--------- HISTOGRAM --------

print("\n")
plt.figure(figsize= (8, 5))

plt.hist(marks, bins= 5, color= "orange", edgecolor= "black")
plt.xlabel("Names")
plt.ylabel("Marks")
plt.title("Marks Distribution - Histogram")
plt.grid(True)
plt.show()

# ------ Scatter Plot -------

print("\n")
plt.figure(figsize= (8, 5))

plt.scatter(names, marks, color= "green", marker= "*", s=80)
plt.xlabel("Names")
plt.ylabel("Marks")
plt.title("Marks of 10 Persons - Scatter Plot")
plt.colorbar(label= "Marks")
plt.grid(True)
# plt.grid(True)
plt.show()

#------PIE CHART -----------
print("\n")
plt.figure(figsize= (8, 5))

plt.pie(marks, labels= names, autopct= "%1.1f%%", startangle= 90)
plt.title("Marks Distribution - Pie Chart")
plt.show()
```

```
import pandas as pd

# Prepare a dataframe having customer name, shopping amount, frequency and city
data= {
    "Customer Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank"],
    "Shopping Amount": [150, 200, 50, 300, 120, 80],
    "Frequency": [5, 3, 10, 2, 7, 4],
    "City": ["New York", "London", "Paris", "Tokyo", "Rome", "Berlin"]
}

df_customers = pd.DataFrame(data)

print("----Customer DataFrame----")
display(df_customers.head())
display(df_customers.tail())

print("\n----Dataset Information----")
print(df_customers.info())

print("\n----Customers with Shopping Amount > 100----")
display(df_customers[df_customers["Shopping Amount"] > 100])

print("\n----Customers with Frequency > 3----")
display(df_customers[df_customers["Frequency"] > 3])
```
