```python
# KNN Regression

from sklearn.neighbors import KNeighborsRegressor

# Dataset: Car Model Year and Price
x = [[1990], [1992], [2000], [2005], [2010], [2012], [2015]]
y = [200000, 220000, 350000, 400000, 500000, 520000, 600000]

# Train KNN Regressor
model = KNeighborsRegressor(n_neighbors=4)
model.fit(x, y)

# Predict price for a 2013 model
prediction = model.predict([[2013]])

print("Predicted Price:", prediction[0])


# Show nearest neighbors
distance, indices = model.kneighbors([[2013]])

print("\nNearest Neighbors:")

for i in range(len(indices[0])):
    index = indices[0][i]
    dist = distance[0][i]

    print(
        "Year:", x[index],
        "Price:", y[index],
        "Distance:", dist
    )
```
