```
# ============================================================
# First Machine Learning Program
# Iris Flower Classification using K-Nearest Neighbors (KNN)
# ============================================================

# Import the Iris dataset that comes built-in with Scikit-learn
from sklearn.datasets import load_iris

# Used to divide our dataset into training data and testing data
from sklearn.model_selection import train_test_split

# K-Nearest Neighbors classification algorithm
from sklearn.neighbors import KNeighborsClassifier

# Used to calculate how accurate our model's predictions are
from sklearn.metrics import accuracy_score


# ------------------------------------------------------------
# 1. LOAD THE DATASET
# ------------------------------------------------------------

# Load the Iris dataset
# The dataset contains measurements of 150 iris flowers
# belonging to 3 different species:
# 0 -> Iris Setosa
# 1 -> Iris Versicolor
# 2 -> Iris Virginica
iris = load_iris()


# ------------------------------------------------------------
# 2. SEPARATE INPUTS (X) AND OUTPUT (Y)
# ------------------------------------------------------------

# X contains the input features used by our ML model.
#
# There are 4 features for each flower:
# 1. Sepal Length
# 2. Sepal Width
# 3. Petal Length
# 4. Petal Width
x = iris.data

# Y contains the correct answer (target/label).
# Our model will learn to predict the species of the flower.
y = iris.target


# ------------------------------------------------------------
# 3. SPLIT DATA INTO TRAINING AND TESTING DATA
# ------------------------------------------------------------

# We divide our dataset into:
#
# Training data -> Used to teach the model
# Testing data  -> Used to check how well the model learned
#
# test_size=0.2 means:
# 80% of the data -> Training
# 20% of the data -> Testing
#
# random_state=42 makes the split reproducible.
# If we run the program again, we get the same split.
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)


# ------------------------------------------------------------
# 4. CREATE THE KNN MODEL
# ------------------------------------------------------------

# Create a K-Nearest Neighbors classifier.
#
# n_neighbors=3 means that when predicting a new flower,
# KNN looks at the 3 closest flowers in the training data
# and uses their classes to decide the prediction.
knn = KNeighborsClassifier(n_neighbors=3)


# ------------------------------------------------------------
# 5. TRAIN THE MODEL
# ------------------------------------------------------------

# Give the training data to the KNN model.
#
# X_train -> Flower measurements
# y_train -> Correct species
#
# The model uses this data to learn the relationship between
# flower measurements and their species.
knn.fit(x_train, y_train)


# ------------------------------------------------------------
# 6. MAKE PREDICTIONS
# ------------------------------------------------------------

# Give the model the TEST data that it has not used for training.
#
# The model predicts the species of each flower.
y_pred = knn.predict(x_test)


# ------------------------------------------------------------
# 7. EVALUATE THE MODEL
# ------------------------------------------------------------

# Compare:
#
# y_test -> Actual/correct answers
# y_pred -> Model's predictions
#
# accuracy_score returns the percentage of predictions
# that were correct.
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy of my first ML model is:", accuracy)
```
