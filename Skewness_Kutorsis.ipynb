```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("penguins")

print("first 5 rows of dataset")
print(df.head())

numeric_data = df.select_dtypes(include="number")

skewness = numeric_data.skew()

print("\nSkewness of numeric columns:")
print(skewness)


kurtosis = numeric_data.kurtosis()


print("\nKurtosis of numeric columns:")
print(kurtosis)

result = pd.DataFrame({"Skewness": skewness, "Kurtosis": kurtosis})
print(result)

for column in numeric_data.columns:
  s=skewness[column]
  k=kurtosis[column]
  print(f"\n{column}")

  if abs(s) < 0.5:
      print("Skewness : approximately symmetric")
  elif s > 0:
      print("Skewness : positively skewed")
  else:
      print("Skewness : negatively skewed")


  if k > 0:
      print("Kurtosis : leptokurtic")
  else:
      print("Kurtosis : platykurtic")
```
