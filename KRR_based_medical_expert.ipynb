```python
# KRR Knowledge reasoning and representing task

# KRR_based_medical_expert

print("=" * 50)
print("Rule based medical expert")
print("=" * 50)

# Collect symptoms
fever = input("Fever (yes/no) :").lower()
cough = input("Cough (yes/no) :").lower()
sneeze = input("Sneeze (yes/no) :").lower()
sore_throat = input("Sore throat (yes/no) :").lower()
headache = input("Headache (yes/no) :").lower()
body_ache = input("Body ache (yes/no) :").lower()
diarrhea = input("Diarrhea (yes/no) :").lower()
loss_of_taste = input("Loss of taste (yes/no) :").lower()
loss_of_smell = input("Loss of smell (yes/no) :").lower()
nausea = input("Nausea (yes/no) :").lower()
runny_nose = input("Runny nose (yes/no) :").lower()
chest_pain = input("Chest pain (yes/no) :").lower()
shortness_of_breath = input("Shortness of breath (yes/no) :").lower()
fatigue = input("Fatigue (yes/no) :").lower()
muscle_weakness = input("Muscle weakness (yes/no) :").lower()
vomiting = input("Vomiting (yes/no) :").lower()
dizziness = input("Dizziness (yes/no) :").lower()
breathlessness = input("Breathlessness (yes/no) :").lower()
wheezing = input("Wheezing (yes/no) :").lower()

print("\n Analyzing the Symptoms .... ...  .......            ...........   ..........       ...........\n")


# =========================================================
# RULES
# =========================================================

# Breathing emergency warning (checked first - most urgent)
if (chest_pain == "yes" and
    (shortness_of_breath == "yes" or breathlessness == "yes")):

    disease = "Urgent respiratory/chest symptoms"


# COVID-19-like symptom pattern
elif (fever == "yes" and cough == "yes" and
    (shortness_of_breath == "yes" or breathlessness == "yes") and
    (loss_of_taste == "yes" or loss_of_smell == "yes")):

    disease = "COVID-19-like respiratory illness"


# Flu-like symptom pattern
elif (fever == "yes" and cough == "yes" and
      headache == "yes" and body_ache == "yes" and
      fatigue == "yes"):

    disease = "Flu-like illness"


# Common cold / upper respiratory pattern
elif (sneeze == "yes" and runny_nose == "yes" and
      sore_throat == "yes" and cough == "yes"):

    disease = "Common cold-like illness"


# Respiratory infection pattern
elif (fever == "yes" and cough == "yes" and
      sore_throat == "yes" and fatigue == "yes"):

    disease = "Respiratory infection"


# Bronchial / airway irritation pattern
elif (cough == "yes" and wheezing == "yes" and
      (shortness_of_breath == "yes" or breathlessness == "yes")):

    disease = "Airway-related respiratory condition"


# Gastrointestinal illness with fever
elif (fever == "yes" and diarrhea == "yes" and
      (nausea == "yes" or vomiting == "yes")):

    disease = "Gastrointestinal infection-like illness"


# Gastrointestinal illness pattern
elif (nausea == "yes" and vomiting == "yes" and
      diarrhea == "yes"):

    disease = "Gastrointestinal illness"


# Migraine/headache-like pattern
elif (headache == "yes" and nausea == "yes" and
      dizziness == "yes"):

    disease = "Migraine-like headache pattern"


# General viral illness pattern
elif (fever == "yes" and fatigue == "yes" and
      muscle_weakness == "yes" and body_ache == "yes"):

    disease = "Viral illness-like pattern"


# Mild upper respiratory illness
elif (cough == "yes" and sore_throat == "yes" and
      runny_nose == "yes"):

    disease = "Upper respiratory illness"


# Chest pain
elif chest_pain == "yes":

    disease = "Chest pain requiring medical evaluation"


# Breathing difficulty
elif (shortness_of_breath == "yes" or
      breathlessness == "yes"):

    disease = "Breathing difficulty requiring medical evaluation"


# Dizziness and weakness
elif (dizziness == "yes" and
      muscle_weakness == "yes"):

    disease = "Weakness/dizziness pattern"


# Only fever
elif fever == "yes":

    disease = "Fever of unspecified cause"


# Only cough
elif cough == "yes":

    disease = "Cough of unspecified cause"


# No significant symptoms
else:
  disease = "No clear symptom pattern identified"


# =========================================================
# RESULT
# =========================================================

print("\nPossible condition:", disease)

# Recommendation rules

print("\nRecommendation:")
if disease == "Urgent respiratory/chest symptoms":
  print(" seek EMERGENCY medical care immediately ")
  print(" do not drive yourself, call for help or have someone take you ")

elif disease == "COVID-19-like respiratory illness":
  print(" consult a doctor immediately ")
  print("ISOLATE yourself")
  print(" consider taking a COVID-19 test ")

elif disease == "Flu-like illness":
  print(" consult a doctor immediately ")
  print(" rest and stay well hydrated ")
  print(" isolate yourself while symptomatic ")

elif disease == "Common cold-like illness":
  print(" rest and stay hydrated ")
  print(" over-the-counter cold remedies may help ")
  print(" consult a doctor if symptoms persist beyond 10 days ")

elif disease == "Respiratory infection":
  print(" consult a doctor for proper evaluation ")
  print(" rest and stay hydrated ")

elif disease == "Airway-related respiratory condition":
  print(" consult a doctor, especially if symptoms worsen ")
  print(" avoid smoke, dust and other respiratory irritants ")

elif disease == "Gastrointestinal infection-like illness":
  print(" stay hydrated with water or oral rehydration solutions ")
  print(" consult a doctor if fever persists or symptoms worsen ")

elif disease == "Gastrointestinal illness":
  print(" stay hydrated with water or oral rehydration solutions ")
  print(" rest and eat light, bland foods ")
  print(" consult a doctor if symptoms persist beyond 2 days ")

elif disease == "Migraine-like headache pattern":
  print(" rest in a quiet, dark room ")
  print(" stay hydrated ")
  print(" consult a doctor if headaches are severe or unusual ")

elif disease == "Viral illness-like pattern":
  print(" rest and stay hydrated ")
  print(" monitor symptoms over the next 24-48 hours ")
  print(" consult a doctor if symptoms worsen ")

elif disease == "Upper respiratory illness":
  print(" rest and stay hydrated ")
  print(" consult a doctor if symptoms worsen or last more than a week ")

elif disease == "Chest pain requiring medical evaluation":
  print(" consult a doctor promptly ")
  print(" seek emergency care if pain is severe or spreading ")

elif disease == "Breathing difficulty requiring medical evaluation":
  print(" consult a doctor promptly ")
  print(" seek emergency care if breathing difficulty is severe or sudden ")

elif disease == "Weakness/dizziness pattern":
  print(" rest and stay hydrated ")
  print(" avoid sudden movements or standing up quickly ")
  print(" consult a doctor if symptoms persist ")

elif disease == "Fever of unspecified cause":
  print(" monitor your temperature regularly ")
  print(" stay hydrated and rest ")
  print(" consult a doctor if fever exceeds 103F/39.4C or lasts more than 3 days ")

elif disease == "Cough of unspecified cause":
  print(" stay hydrated and rest ")
  print(" consult a doctor if the cough persists beyond 2-3 weeks ")

else:
  print(" no strong pattern detected ")
  print(" if you are feeling unwell, consult a doctor for proper evaluation ")

print("\nNote: This tool provides general guidance only and is not a substitute for professional medical advice, diagnosis, or treatment.")
```
