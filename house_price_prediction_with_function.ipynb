```python
# Import NumPy for working with arrays and numerical calculations
import numpy as np

# Import Matplotlib for creating graphs and visualizations
import matplotlib.pyplot as plt


# ---------------------------------------------------------
# TRAINING DATA
# ---------------------------------------------------------

# x_train represents the size of houses
# Unit: 1000 square feet
#
# Example:
# 1.0 means 1000 sqft
# 2.0 means 2000 sqft
# 5.0 means 5000 sqft

x_train = np.array([
    0.5, 0.8, 1.0, 1.2, 1.5,
    1.7, 2.0, 2.2, 2.5, 2.7,
    3.0, 3.2, 3.5, 3.7, 4.0,
    4.2, 4.5, 4.7, 5.0, 5.5,
    6.0, 6.5, 7.0, 7.5, 8.0
])


# y_train represents the actual price of each house
# Unit: 1000s of dollars
#
# Example:
# 150 means $150,000
# 300 means $300,000
# 500 means $500,000

y_train = np.array([
     90, 120, 150, 165, 180,
    195, 210, 225, 250, 270,
    290, 305, 320, 335, 350,
    370, 390, 410, 430, 460,
    490, 520, 550, 580, 620
])


# ---------------------------------------------------------
# MODEL PARAMETERS
# ---------------------------------------------------------

# w = weight/slope of our linear model
# It determines how much the predicted price changes
# when the house size increases by 1 unit

w = 50


# b = bias/intercept
# It is the starting value of our prediction
# when x = 0

b = 100


# ---------------------------------------------------------
# MODEL FUNCTION
# ---------------------------------------------------------

# This function calculates the predicted house prices
#
# Formula of our linear model:
#
# f(x) = w * x + b
#
# x = house size
# w = weight/slope
# b = bias/intercept
#
# The function returns the predicted prices

def compute_model_output(x, w, b):

    # Get the number of training examples
    #
    # x.shape[0] gives the number of elements in x
    m = x.shape[0]


    # Create an array filled with zeros
    # This array will store our predictions
    f_wb = np.zeros(m)


    # Loop through every training example
    for i in range(m):

        # Calculate the prediction using:
        #
        # f(x) = w * x + b
        #
        # Example:
        # if x = 2
        # w = 50
        # b = 100
        #
        # prediction = 50 * 2 + 100
        #            = 200
        #
        # Since price is in thousands,
        # 200 means $200,000

        f_wb[i] = w * x[i] + b


    # Return all calculated predictions
    return f_wb


# ---------------------------------------------------------
# MAKE PREDICTIONS
# ---------------------------------------------------------

# Call our function to calculate predicted prices
# for all houses in x_train

tmp_f_wb = compute_model_output(x_train, w, b)


# ---------------------------------------------------------
# PLOT THE MODEL PREDICTION
# ---------------------------------------------------------

# Plot the predicted values as a line
#
# x_train -> x-axis
# tmp_f_wb -> predicted values on y-axis
#
# c='b' means blue color
# label gives a name to this line

plt.plot(
    x_train,
    tmp_f_wb,
    c='b',
    label='Our Prediction'
)


# ---------------------------------------------------------
# PLOT ACTUAL DATA
# ---------------------------------------------------------

# Plot the actual training data as points
#
# x_train -> house sizes
# y_train -> actual prices
#
# marker='x' means use an X-shaped marker
# c='r' means red color

plt.scatter(
    x_train,
    y_train,
    marker='x',
    c='r',
    label='Actual Values'
)


# ---------------------------------------------------------
# GRAPH LABELS
# ---------------------------------------------------------

# Set the title of the graph
plt.title("House Price Prediction With Function")


# Label for the Y-axis
plt.ylabel("Price (in 1000s of dollars)")


# Label for the X-axis
plt.xlabel("Size (1000 sqft)")


# Display the labels for our prediction
# and actual values

plt.legend()


# Display the final graph
plt.show()
```
